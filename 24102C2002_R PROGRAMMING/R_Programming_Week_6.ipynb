## 1. For Loop

```r
# Syntax:
# for (name in vector) {
#   commands to be executed
# }

for (i in 1:5) {
  print(i^2)
}
```

```r
# For loop with a vector

for (i in c(2, 4, 6, 7)) {
  print(i^2)
}
```

```r
# Example: count values satisfying a condition

x <- c(2, 4, 6, 8, 10, 12)

excount <- function(x) {
  count <- 0

  for (xval in x) {
    if (xval / 2 > 3) {
      count <- count + 1
    }
  }

  print(count)
}

excount(x)
```

## 2. Nested For Loops

```r
child <- c("child1", "child2", "child3")
sweet <- c("sweet1", "sweet2", "sweet3")

for (x in child) {
  for (y in sweet) {
    print(paste(x, y))
  }
}
```

## 3. The `break` Command

```r
drink <- c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "tea") {
    break
  }

  print(x)
}
```

## 4. The `next` Command

```r
# Skip lemonade

drink <- c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "lemonade") {
    next
  }

  print(x)
}
```

```r
# Skip tea

drink <- c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "tea") {
    next
  }

  print(x)
}
```

## 5. While Loop

```r
# Syntax:
# while (condition) {
#   commands to be executed as long as condition is TRUE
# }

i <- 1

while (i < 10) {
  print(i^2)
  i <- i + 2
}
```

```r
# Example using readline()
# Enter a number less than or equal to 25 when prompted.

sumfunction <- function() {
  sum <- 0

  number <- as.integer(
    readline(prompt = "Please select any number less than 25: ")
  )

  while (number <= 25) {
    sum <- sum + number
    number <- number + 1
  }

  print(paste(
    "The sum of numbers received from the While Loop: ",
    sum
  ))
}

sumfunction()
```

## 6. Repeat Loop

```r
# The repeat loop continues until break is executed.

i <- 1

repeat {
  print(i^2)
  i <- i + 2

  if (i > 10) {
    break
  }
}
```

```r
# Example using next and break

i <- 1

repeat {
  i <- i + 1

  if (i < 10) {
    next
  }

  print(i^2)

  if (i >= 13) {
    break
  }
}
```

## 7. Functions

```r
# A function with one argument

abc <- function(x) {
  x^2
}

abc(3)
abc(6)
abc(9)
```

```r
# A function with two arguments

abc <- function(x, y) {
  x^2 + y^2
}

abc(3, 4)
abc(10, 10)
abc(-2, -3)
```

```r
# Another function example

abc <- function(x) {
  sin(x)^2 + cos(x)^2 + x
}

abc(9)
abc(99)
abc(-15)
```

```r
# Function without arguments

abc <- function() {
  for (i in 1:3) {
    print(i^3)
  }
}

abc()
```

## 9. Sequences Using `seq()`

```r
# Default increment (+1)

seq(from = 2, to = 4)
seq(from = 4, to = 2)
seq(from = -4, to = 4)
```

```r
# Sequence with constant increment

seq(from = 10, to = 20, by = 2)
```

```r
# Sequence with constant decrement

seq(from = 20, to = 10, by = -2)
```

```r
# Fractional decrement

seq(from = 3, to = -2, by = -0.5)
```

```r
# Predefined length with default increment

seq(to = 10, length.out = 10)
seq(from = 10, length.out = 10)
```

```r
# Predefined length with fractional increment

seq(from = 10, length.out = 10, by = 0.1)
```

```r
# Predefined length with constant decrement

seq(from = 10, length.out = 10, by = -2)
```

```r
# Predefined length with fractional decrement

seq(from = 10, length.out = 5, by = -0.2)
```

Cell 1 — Lists: Create and Merge

```r
# Create two lists
list1 = list(1, 2, 3)
list2 = list("water", "juice", "lemonade")

# Display lists
list1
list2

# Merge two lists
list12 = c(list1, list2)

# Display merged list
list12
```

Cell 2 — Convert List to Vector

```r
# Create lists
list1 = list(1, 2, 3)
list2 = list("water", "juice", "lemonade")

# Convert list to vector
unlist(list1)
unlist(list2)

# Check modes
mode(list1)
mode(unlist(list1))
```

Cell 3 — Append to List

```r
list1 = list(1, 2, 3)
list2 = list("water", "juice", "lemonade")

# Append at the end
append(list1, 100)
append(list2, "coffee")
```

Cell 4 — Append After a Position

```r
list1 = list(1, 2, 3)
list2 = list("water", "juice", "lemonade")

# Append after position 2
append(list1, 100, after = 2)
append(list2, "coffee", after = 2)
```

Cell 5 — Remove from List

```r
list1 = list(1, 2, 3)
list2 = list("water", "juice", "lemonade")

# Remove element at position 2
list1[-2]

# Remove element at position 1
list2[-1]
```

Cell 6 — Extract Elements from List

```r
list1 = list(1, 2, 3, 4, 5, 6)
list2 = list("water", "juice", "lemonade", "tea", "coffee", "milk")

# Extract positions 2 to 4
list1[2:4]

# Extract positions 1, 3 and 5
list1[c(1, 3, 5)]

# Extract from list2
list2[2:4]
list2[c(1, 3, 5)]
```

#Vector Indexing
Cell 7 — Basic Vector Indexing

```r
# Create vector
x = 1:10

# Display vector
x

# Values greater than 5
x[(x > 5)]

# Even values
x[(x %% 2 == 0)]

# Odd values
x[(x %% 2 == 1)]
```

Cell 8 — Missing Values

```r
# Create vector
x = 1:10

# Replace 5 with NA
x[5] = NA

# Display vector
x

# Remove missing value
y = x[!is.na(x)]

# Display y
y

# Calculate mean
mean(x)
mean(y)
```

Cell 9 — Negative Indexing

```r
x = 1:10

# Remove first five elements
x[-(1:5)]

# Same result
x[6:10]
```

#Factors
Cell 15 — Create a Factor

```r
# Create vector
x = c(3, 4, 5, 6, 1, 2, 3, 3, 4, 4, 5, 6)

# Convert to factor
y = as.factor(x)

# Display
x
y
```

Cell 16 — Factor with Character Values

```r
x = factor(c("lemonade", "lemonade", "juice", "lemonade", "water"))

# Display factor
x
```

Cell 17 — factor() with Levels and Labels

```r
# Die outcomes
y = c(1, 4, 3, 5, 4, 2, 4)

# Possible values
possible.dieface = c(1, 2, 3, 4, 5, 6)

# Labels
labels.dieface = c(
  "one", "two", "three",
  "four", "five", "six"
)

# Create factor
facy = factor(
  y,
  levels = possible.dieface,
  labels = labels.dieface
)

# Display
facy
```

Cell 18 — class()

```r
# Check class of different objects
class(9)
class("9")
class(print)

# Create matrix
x = matrix(
  nrow = 2,
  ncol = 2,
  data = 1:4
)

# Check class
class(x)
```

Cell 19 — unclass()

```r
# Create factor
brands = c("A", "A", "B", "B", "B", "B", "C")

brands_fac = factor(brands)

# Display
brands
brands_fac

# Remove factor class temporarily
unclass(brands_fac)
```

Cell 20 — Convert Factor Codes to Values

```r
brands = c("A", "A", "B", "B", "B", "B", "C")

brands_fac = factor(brands)

# Create colours
colours = c("blue", "green", "red")

# Convert factor codes to colours
colours[unclass(brands_fac)]
```

Cell 21 — Control Factor Level Order

```r
x = factor(
  c("lemonade", "lemonade", "juice", "lemonade", "water"),
  levels = c("water", "juice", "lemonade")
)

# Display
x

# Show numeric codes
unclass(x)

# Show levels
levels(x)
```

Cell 22 — Ordered Factor

```r
income = ordered(
  c("high", "high", "low", "medium", "medium"),
  levels = c("low", "medium", "high")
)

# Display
income

# Show underlying codes
unclass(income)
```

Cell 26 — justify

```r
# Centre justification
format(
  c("A", "BB", "CCC", "DDDD"),
  width = 7,
  justify = "centre"
)

# Left justification
format(
  c("A", "BB", "CCC", "DDDD"),
  width = 7,
  justify = "left"
)

# Right justification
format(
  c("A", "BB", "CCC", "DDDD"),
  width = 7,
  justify = "right"
)

# No justification
format(
  c("A", "BB", "CCC", "DDDD"),
  width = 7,
  justify = "none"
)
```

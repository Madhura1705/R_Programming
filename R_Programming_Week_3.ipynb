# Foundations of R - Lecture Notes Week 03
Basics of Calculations and Matrices

```r
############################################################
# 1. Vector Arithmetic
############################################################

# Addition
c(2,3,5,7) + c(-2,-3,-5,8)

# Subtraction
c(2,3,5,7) - c(-2,-3,-5,8)

# Multiplication
c(2,3,5,7) * c(-2,-3,-5,8)

# Division
c(2,3,5,7) / c(-2,-3,-5,8)
```

```r
############################################################
# 2. Power Operations
############################################################

# Scalar power
2^3
2**3

# Square root using power
2^0.5
2**0.5

# Negative power
2^-0.5

# Vector squared
c(2,3,5,7)^2

# Vector raised to another vector
c(2,3,5,7)^c(2,3)

# Another vector example
c(1,2,3,4,5,6)^c(2,3,4)

# Different length vectors (Warning)
c(2,3,5,7)^c(2,3,4)
```

```r
############################################################
# 3. Integer Division
############################################################

2 %/% 2
5 %/% 2
7 %/% 3

# Vector with scalar
c(2,3,5,7) %/% 2

# Vector with vector
c(2,3,5,7) %/% c(2,3)

# Warning example
c(2,3,5) %/% c(2,3)
```

```r
############################################################
# 4. Modulo Division
############################################################

2 %% 2
3 %% 2
7 %% 3
7 %% 4

# Vector with scalar
c(2,3,5,7) %% 2

# Vector with vector
c(2,3,5,7) %% c(2,3)

# Warning example
c(2,3,5) %% c(2,3)
```

```r
############################################################
# 5. Built-in Functions
############################################################

# Maximum
max(1.2,3.4,-7.8)
max(c(1.2,3.4,-7.8))

# Minimum
min(1.2,3.4,-7.8)
min(c(1.2,3.4,-7.8))

# Mean
mean(c(2,3,4))
```

```r
############################################################
# 6. Mathematical Functions
############################################################

# Absolute value
abs(-4)
abs(c(-1,-2,-3,4,5))

# Square root
sqrt(4)
sqrt(c(4,9,16,25))

# Sum
sum(c(2,3,5,7))

# Product
prod(c(2,3,5,7))

# Round
round(1.23)
round(1.83)

# Natural logarithm
log(10)
log(exp(1))
log(c(10,100,1000))

# Base-10 logarithm
log10(10)
log10(100)
log10(c(10,100,1000))
```

```r
############################################################
# 7. Variable Assignment
############################################################

x1 <- c(1,2,3,4)
x1

x2 <- x1^2
x2
```

```r
############################################################
# 8. Combining Operations
############################################################

c(1,2,3,4) + sum(c(1,2,3,4))*prod(c(1,2))

abs(c(1,2,3,4) - sum(c(1,2,3,4))*prod(c(1,2)))
```

```r
############################################################
# 9. Matrix Creation
############################################################

# Matrix filled column-wise (default)
x <- matrix(
  data=c(1,2,3,4,5,6,7,8),
  nrow=4,
  ncol=2
)

x
```

```r
############################################################
# 10. Access Matrix Element
############################################################

x[3,2]
```

```r
############################################################
# 11. Matrix by Column
############################################################

x <- matrix(
  data=c(1,2,3,4,5,6,7,8),
  nrow=4,
  ncol=2,
  byrow=FALSE
)

x
```

```r
############################################################
# 12. Matrix by Row
############################################################

y <- matrix(
  data=c(1,2,3,4,5,6,7,8),
  nrow=4,
  ncol=2,
  byrow=TRUE
)

y
```

```r
############################################################
# 13. Matrix Properties
############################################################

# Dimensions
dim(x)

# Number of rows
nrow(x)

# Number of columns
ncol(x)

# Storage mode
mode(x)

# Matrix attributes
attributes(x)
```

```r
############################################################
# 14. Help
############################################################

help("matrix")
```

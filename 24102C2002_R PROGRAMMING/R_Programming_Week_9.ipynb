#Lecture 36 — cat() and String Formatting

Cell 1 — print() vs cat()

```r
# Using print()
print("The zero occurs at")
print(2*pi)
print("radians")

# Using cat()
cat("The zero occurs at", 2*pi, "radians.", "\n")
```

Cell 2 — Date with cat()

```r
d <- date()

cat("Today's date is:", d, "\n")
```

Cell 3 — sep in cat()

```r
x <- 1:10

cat(x, sep = " ++ ")
cat("\n")

cat(x, sep = " / ")
cat("\n")
```

Cell 4 — Square of a number

```r
x <- 7

cat("The square of", x, "is", x^2, "!\n")
```

Cell 5 — Square root and format()

```r
x <- 7

cat(
  "The square root of",
  x,
  "is approximately",
  format(sqrt(x), digits = 3),
  "\n"
)
```

Cell 6 — Printing a vector

```r
evenno <- c(2, 4, 6, 8, 10)

cat("The first few even numbers are:", evenno, "...\n")
```

Cell 7 — fill and labels

```r
x <- 1:10

cat(
  x,
  fill = 2,
  labels = paste("(", letters[1:10], "):")
)
```

#Lecture 37 — paste() Function

Cell 8 — Basic paste()

```r
paste("Everybody", "loves", "R Programming.")
```

Cell 9 — sep

```r
paste("Everybody", "loves", "R Programming.", sep = "*")

paste("Everybody", "loves", "R Programming.", sep = "===")
```

Cell 10 — paste() and as.character()

```r
paste(1:12)

as.character(1:12)
```

Cell 11 — Vector with paste()

```r
names <- c(
  "Prof. Singh",
  "Mr. Venkat",
  "Dr. Jha"
)

names

paste(names, "is", "a good", "person.")
```

Cell 12 — collapse

```r
names <- c(
  "Prof. Singh",
  "Mr. Venkat",
  "Dr. Jha"
)

paste(
  names,
  "is",
  "a good",
  "person.",
  collapse = ", and "
)
```

#Lecture 38 — strsplit()

Cell 19 — Split using !

```r
x <- "The&!syntax&!of&!paste&!is!&available!&inthe online-help"

x

strsplit(x, split = "!")
```

Cell 20 — Split using "&!"

```r
x <- "The&!syntax&!of&!paste&!is!&available!&inthe online-help"

strsplit(x, split = "&!")
```

Cell 21 — Access elements of strsplit()

```r
x <- "The&!syntax&!of&!paste&!is!&available!&inthe online-help"

y <- strsplit(x, split = "!&")

y

y[[1]][1]
y[[1]][2]
y[[1]][3]
```

#Date splitting and Matrix

Cell 22 — Create dates

```r
dates <- c(
  "2020-07-24",
  "2021-08-25",
  "2022-09-26",
  "2023-10-27"
)

dates
```

Cell 23 — Split dates

```r
datesplt <- strsplit(dates, "-")

datesplt
```

Cell 24 — Convert to character matrix

```r
datemat <- matrix(
  unlist(datesplt),
  nrow = 4,
  ncol = 3,
  byrow = TRUE
)

datemat
```

Cell 25 — Convert to numeric matrix

```r
datematrix <- matrix(
  as.numeric(unlist(datesplt)),
  nrow = 4,
  ncol = 3,
  byrow = TRUE
)

datematrix
```

Cell 26 — Split every character

```r
strsplit("Shalabh", split = "")
```

#Lecture 39 — Counting Characters

Cell 27 — nchar()

nchar() returns the number of characters in each element of a character vector.

```r
x <- "R course 24.07.2022"
y <- "Number of participants: 25"

nchar(x)
nchar(y)
```

Cell 28 — nzchar()

```r
x <- "R course 24.07.2022"
y <- "Number of participants: 25"

nzchar(x)
nzchar(y)
```

Cell 29 — nzchar() with empty string

```r
x <- c("Apple", "Banana", "Cake")

x
nzchar(x)

y <- c("Apple", "", "Cake")

y
nzchar(y)
```

nchar() with vectors and numbers

```r
x <- c("Apple", "Banana", "Cake")

nchar(x)
```

```r
y <- c(2, 4, 6)

nchar(y)
```

```r
z <- c(11, 222, 3333)

nchar(z)
```

Logical Operators

22. | and ||

```r
x <- 8

(x < 10) || (x < 2)

x <- 18

(x < 10) || (x < 2)
```

23. | vs || with Vector

```r
# | works element-wise on vectors
x <- c(8,18)

(x < 10) | (x < 2)
```

24. & and &&

```r
# && works with a single value

x <- 5
(x < 10) && (x > 2)

x <- 15
(x < 10) && (x > 2)


# & works element-wise with vectors

x <- c(8,18)

(x < 10) & (x > 2)
```

25. Filtering Values Using Logical Conditions

```r
x <- 1:6

(x > 2) & (x < 5)

x[(x > 2) & (x < 5)]
```

26. OR Condition

```r
x <- 1:6

(x > 2) | (x < 5)

x[(x > 2) | (x < 5)]
```

27. && Checks First Element

```r
# && works with single logical values

x <- 1:6

# Element-wise comparison
(x > 2) & (x < 5)

# Demonstrating && using only the first element
(x[1] > 2) && (x[1] < 5)
```

28. Standard Logical Operations

```r
x <- TRUE
y <- FALSE

x & y

x | y

!x
```

29. is.logical()

```r
x <- 5

Logical1 <- (x > 2)
Logical1
is.logical(Logical1)

Logical2 <- (x < 10)
Logical2
is.logical(Logical2)

Logical3 <- (x != 5)
Logical3
is.logical(Logical3)
```

30. More Logical Expressions

```r
x <- 5

Logical4 <- (2*x > 11)
Logical4
is.logical(Logical4)

Logical5 <- (3*x < 20)
Logical5
is.logical(Logical5)
```

#PART 2 — IMPORTING EXCEL DATA

Create sample Excel file

Install and load readxl

```python
import pandas as pd

sheet1 = pd.DataFrame({
    "Variable 1": [1, 2, 3, 4, 5],
    "Variable 2": [10, 20, 30, 40, 50],
    "Variable 3": [100, 200, 300, 400, 500]
})

sheet2 = pd.DataFrame({
    "Variable 4": [6, 7, 8, 9, 10],
    "Variable 5": [110, 120, 130, 140, 150],
    "Variable 6": [110, 210, 310, 410, 510]
})

with pd.ExcelWriter("spexcel.xlsx") as writer:
    sheet1.to_excel(writer, sheet_name="Sheet1", index=False)
    sheet2.to_excel(writer, sheet_name="Sheet2", index=False)

print("spexcel.xlsx created successfully")
```

#PART 13 — DIRECTION BAR PLOTS

Direct direction barplot

```python
barplot(direction)
```

Frequency barplot

```python
barplot(table(direction))
```

Relative-frequency barplot

```python
barplot(table(direction) / length(direction))
```

Add colours

```python
barplot(
  table(direction),
  col = c("red", "green", "blue")
)
```

Add title

```python
barplot(
  table(direction),
  col = c("red", "green", "blue"),
  main = "Directions of food delivery"
)
```

Add legend

```python
barplot(
  table(direction),
  col = c("red", "green", "blue"),
  main = "Directions of food delivery",
  legend.text = c("dir1", "dir2", "dir3")
)
```

Add subtitle

```python
barplot(
  table(direction),
  col = c("red", "green", "blue"),
  main = "Directions of food delivery",
  legend.text = c("dir1", "dir2", "dir3"),
  sub = "Three directions"
)
```

FINAL BARPLOT

```python
barplot(
  table(direction),
  col = c("red", "green", "blue"),
  main = "Directions of food delivery",
  legend.text = c("dir1", "dir2", "dir3"),
  sub = "Three directions",
  xlab = "Food Delivery Directions",
  ylab = "Number of Deliveries"
)
```
